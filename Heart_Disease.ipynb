## ***SQL queries***

```python
df.columns
```

```python
query1 = """ SELECT "Patient_ID", "Age", "Gender", "HeartDisease" FROM patients WHERE "Age" > 60;"""

result1 = pd.read_sql(query1, engine)
print(result1.head(5))
print("Total patients over 60:", len(result1))
```

```python
query2 =""" SELECT "HeartDisease", COUNT(*) AS "total_patients", AVG("TotalChol") AS "avg_cholesterol" FROM patients GROUP BY "HeartDisease";"""
result2 = pd.read_sql(query2, engine)
print(result2)
```

```python
query3 = """ SELECT "Patient_ID", "Age", CASE WHEN "Age" < 30 THEN 'Young'
WHEN "Age" BETWEEN 30 AND 60 THEN 'Middle-aged'
ELSE 'Senior' END AS "Age_group"
FROM patients LIMIT 10;"""

result3 = pd.read_sql(query3, engine)
print(result3)
```

```python
query4 = """ SELECT "Patient_ID", "Age", "TotalChol" FROM patients WHERE "TotalChol" > (SELECT AVG("TotalChol") FROM patients); """
result4 = pd.read_sql(query4, engine)
print(result4.head(5))
print("Patients with above-average cholesterol:", len(result4))
```
